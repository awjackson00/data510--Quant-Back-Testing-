```python
# ARIMA Time-Series Modeling
# Fit ARIMA models to portfolio-level return series and evaluate
# forecast accuracy.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
# Build combined_returns for ARIMA
# Purpose:
# Combine student portfolios and benchmarks into one dataframe

import pandas as pd

returns_to_combine = []

if "student_portfolio_returns" in globals():
    returns_to_combine.append(student_portfolio_returns)
    print("Added student_portfolio_returns:", student_portfolio_returns.shape)
else:
    print("student_portfolio_returns not found")

if "benchmark_portfolio_returns" in globals():
    returns_to_combine.append(benchmark_portfolio_returns)
    print("Added benchmark_portfolio_returns:", benchmark_portfolio_returns.shape)
else:
    print("benchmark_portfolio_returns not found")

if "individual_benchmark_returns" in globals():
    returns_to_combine.append(individual_benchmark_returns)
    print("Added individual_benchmark_returns:", individual_benchmark_returns.shape)
else:
    print("individual_benchmark_returns not found")

if len(returns_to_combine) == 0:
    raise NameError("No return dataframes found. Run the notebook cells that create portfolio returns first.")

combined_returns = pd.concat(returns_to_combine, axis=1)

# Remove duplicated columns if any exist
combined_returns = combined_returns.loc[:, ~combined_returns.columns.duplicated()]

# Drop rows where every column is missing
combined_returns = combined_returns.dropna(how="all")

print("\ncombined_returns created successfully")
print("Shape:", combined_returns.shape)
print("Date range:", combined_returns.index.min(), "to", combined_returns.index.max())

display(combined_returns.head())
```

## 1. Choose one portfolio or benchmark series

```python
# Check which expected dataframes exist
possible_names = [
    "combined_returns",
    "student_portfolio_returns",
    "benchmark_portfolio_returns",
    "individual_benchmark_returns",
    "portfolio_returns",
    "portfolio_returns_tidy",
    "asset_returns"
]

for name in possible_names:
    if name in globals():
        obj = globals()[name]
        print(f"{name}: FOUND, shape = {obj.shape}")
    else:
        print(f"{name}: NOT FOUND")
```

```python
# Change this to one of the columns in combined_returns
target_series_name = combined_returns.columns[0]

series = combined_returns[target_series_name].dropna()

print("Selected series:", target_series_name)
print("Start date:", series.index.min())
print("End date:", series.index.max())
print("Observations:", len(series))
```

## 2. Train-test split

```python
train_size = int(len(series) * 0.8)

train = series.iloc[:train_size]
test = series.iloc[train_size:]

print("Train observations:", len(train))
print("Test observations:", len(test))
```

## 3. Fit ARIMA model

```python
# ARIMA(p,d,q)
# p = autoregressive terms
# d = differencing
# q = moving average terms
#
# For daily returns, d=0 is usually reasonable because returns
# are often already stationary. We start with ARIMA(1,0,1).

model = ARIMA(train, order=(1, 0, 1))
model_fit = model.fit()

print(model_fit.summary())
```

## 4. Forecast the test period

```python
forecast = model_fit.forecast(steps=len(test))
forecast.index = test.index
```

## 5. Evaluate forecast accuracy

```python
mae = mean_absolute_error(test, forecast)
rmse = np.sqrt(mean_squared_error(test, forecast))

directional_accuracy = np.mean(np.sign(test) == np.sign(forecast))

print("MAE:", mae)
print("RMSE:", rmse)
print("Directional Accuracy:", directional_accuracy)
```

## 6. Plot actual vs forecasted returns

```python
plt.figure(figsize=(12, 6))
plt.plot(test.index, test, label="Actual Returns")
plt.plot(forecast.index, forecast, label="ARIMA Forecast")
plt.title(f"ARIMA Forecast: {target_series_name}")
plt.xlabel("Date")
plt.ylabel("Daily Return")
plt.legend()
plt.show()
```

```python

# ARIMA Actual vs Forecast Plots for Each Portfolio
# Creates one plot per portfolio/benchmark

from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt

for column in combined_returns.columns:
    series = combined_returns[column].dropna()

    if len(series) < 100:
        print(f"Skipping {column}: not enough observations")
        continue

    train_size = int(len(series) * 0.8)
    train = series.iloc[:train_size]
    test = series.iloc[train_size:]

    model = ARIMA(train, order=(1, 0, 1))
    model_fit = model.fit()

    forecast = model_fit.forecast(steps=len(test))
    forecast.index = test.index

    plt.figure(figsize=(12, 5))
    plt.plot(test.index, test, label="Actual Returns")
    plt.plot(forecast.index, forecast, label="ARIMA Forecast")

    plt.title(f"ARIMA Forecast: {column}")
    plt.xlabel("Date")
    plt.ylabel("Daily Return")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
